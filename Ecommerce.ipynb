### 3. Unindo as tabelas

```python
import pandas as pd

df_orders = pd.read_csv("./olistbr/olist_orders_dataset.csv") 
df_customers = pd.read_csv("./olistbr/olist_customers_dataset.csv") 
df_items = pd.read_csv("./olistbr/olist_order_items_dataset.csv")
df_payments = pd.read_csv("./olistbr/olist_order_payments_dataset.csv") 
df_reviews = pd.read_csv("./olistbr/olist_order_reviews_dataset.csv") 
df_products = pd.read_csv("./olistbr/olist_products_dataset.csv") 
df_sellers = pd.read_csv("./olistbr/olist_sellers_dataset.csv") 
```

```python
df_full_table = pd.merge(df_orders, df_customers, on="customer_id", how="left")

df_full_table = pd.merge(df_full_table, df_items, on="order_id", how="left")

df_full_table = pd.merge(df_full_table, df_payments, on="order_id", how="left")

df_full_table = pd.merge(df_full_table, df_reviews, on="order_id", how="left")

df_full_table = pd.merge(df_full_table, df_sellers, on= "seller_id", how="left")

df_full_table = pd.merge(df_full_table, df_products, on="product_id", how="left")

print(df_full_table.info())
```

```python
# Célula de Código
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configuração de estilo para os gráficos
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

# Unindo a tabela de produtos que faltou no merge inicial para termos as categorias
df_full_table = pd.merge(df_full_table, df_products, on="product_id", how="left")

# Convertendo colunas de data para o formato datetime
df_full_table['order_purchase_timestamp'] = pd.to_datetime(df_full_table['order_purchase_timestamp'])
df_full_table['order_delivered_customer_date'] = pd.to_datetime(df_full_table['order_delivered_customer_date'])
df_full_table['order_estimated_delivery_date'] = pd.to_datetime(df_full_table['order_estimated_delivery_date'])
```

Para entender a logística do e-commerce, precisamos calcular o tempo real de entrega (em dias) e compará-lo com o tempo estimado. Isso nos ajudará a identificar quais estados sofrem com as maiores demoras e representam gargalos logísticos.

```python
# Calculando o tempo de entrega em dias
df_full_table['tempo_entrega_dias'] = (df_full_table['order_delivered_customer_date'] - df_full_table['order_purchase_timestamp']).dt.days

# Calculando o atraso (se a entrega foi feita depois da estimativa)
df_full_table['atraso_dias'] = (df_full_table['order_delivered_customer_date'] - df_full_table['order_estimated_delivery_date']).dt.days
df_full_table['foi_atrasado'] = df_full_table['atraso_dias'] > 0

# Agrupando por estado do cliente para ver a média de dias de entrega
tempo_por_estado = df_full_table.groupby('customer_state')['tempo_entrega_dias'].mean().sort_values(ascending=False).reset_index()

# Visualização: Tempo médio de entrega por estado
plt.figure(figsize=(14, 6))
sns.barplot(data=tempo_por_estado, x='customer_state', y='tempo_entrega_dias', palette='magma')
plt.title('Gargalo Logístico: Tempo Médio de Entrega por Estado (em dias)', fontsize=16)
plt.xlabel('Estado do Cliente', fontsize=12)
plt.ylabel('Média de Dias para Entrega', fontsize=12)
plt.axhline(df_full_table['tempo_entrega_dias'].mean(), color='red', linestyle='--', label='Média Nacional')
plt.legend()
plt.show()
```

Com a logística mapeada, olhamos para o comportamento de consumo. Quais são as categorias de produtos que movem o maior volume de vendas da plataforma?

```python
# Contando as categorias de produtos mais vendidas
top_categorias = df_full_table['product_category_name'].value_counts().head(10).reset_index()
top_categorias.columns = ['Categoria', 'Quantidade de Vendas']

# Visualização: Top 10 Categorias
plt.figure(figsize=(12, 6))
sns.barplot(data=top_categorias, y='Categoria', x='Quantidade de Vendas', palette='viridis')
plt.title('Top 10 Categorias de Produtos Mais Vendidas', fontsize=16)
plt.xlabel('Volume de Pedidos', fontsize=12)
plt.ylabel('Categoria do Produto', fontsize=12)
plt.show()
```

A demora na entrega afeta diretamente a nota que o cliente dá para a compra? Vamos cruzar o tempo de entrega com o `review_score` para validar essa hipótese.

```python
# Agrupando a nota de avaliação (review_score) pelo tempo médio de entrega
satisfacao_entrega = df_full_table.dropna(subset=['review_score', 'tempo_entrega_dias'])
media_entrega_por_nota = satisfacao_entrega.groupby('review_score')['tempo_entrega_dias'].mean().reset_index()

# Visualização: Impacto da entrega na nota do cliente
plt.figure(figsize=(10, 5))
sns.lineplot(data=media_entrega_por_nota, x='review_score', y='tempo_entrega_dias', marker='o', color='teal', linewidth=2.5)
plt.title('Impacto do Tempo de Entrega na Satisfação do Cliente', fontsize=16)
plt.xlabel('Nota de Avaliação (1 a 5)', fontsize=12)
plt.ylabel('Tempo Médio de Entrega (Dias)', fontsize=12)
plt.gca().invert_xaxis() # Invertendo o eixo X para mostrar que quanto menor a nota, maior a demora
plt.show()
```
